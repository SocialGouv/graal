```python
import os

# Usage example
DATA_FOLDER = os.getenv("DATA_FOLDER")
file_path = f"{DATA_FOLDER}/indications_flechages_2024.csv"
sheet_name = "Sheet1"
output_path = f"{DATA_FOLDER}/cleaned_{sheet_name}.csv"
```

```python
import pandas as pd

df = pd.read_csv(file_path, encoding="utf-8")
display(df)
```

```python
empty_ratio = df.isnull().sum() / len(df)
print(empty_ratio)
```

```python
df.loc[
    (df["Personne en charge"].isnull()) & (~df["Sujets pris en charge"].isnull()),
    "Personne en charge",
] = df["Bureaux"]
display(df)
```

```python
df.to_csv(output_path, index=False, encoding="utf-8-sig")
```
