```python
import json
import pandas as pd
from amendements_intelligents.clients.summary_generator_clients import (
    SummaryGeneratorOllamaClient,
)

DATA_FOLDER = os.getenv("DATA_FOLDER")
file_name = "PLFSS 2024"
file_path = f"{DATA_FOLDER}/{file_name}.json"

with open(file_path, "r") as file:
    data = json.load(file)

    # Create a dataframe with the 'data' variable
    df = pd.DataFrame(data)

    # Perform exploratory analysis
    # Get the number of rows and columns
    num_rows, num_cols = df.shape
    print(f"Number of rows: {num_rows}")
    print(f"Number of columns: {num_cols}")

    # Get the column names
    column_names = df.columns.tolist()
    print("Column names:")
    print(column_names)

    # Calculate the ratio of missing values for each column
    missing_ratio = df.isnull().mean()

    # Print the missing ratio for each column
    print("Missing value ratio for each column:")
    print(missing_ratio * 100)

    # Calculate the length of each value in the 'Exposé des motifs' column
    lengths = df["Exposé des motifs"].dropna().apply(len)

    # Display the minimum, mean, and maximum length
    print("Analysis of 'Exposé des motifs':")
    print(f"Minimum length: {lengths.min()}")
    print(f"Mean length: {lengths.mean()}")
    print(f"Maximum length: {lengths.max()}")

    # Filter rows where 'Exposé des motifs' is not empty
    filtered_df = df[df["Exposé des motifs"].notnull()]

    # Calculate the mean and standard deviation of the length of 'Exposé des motifs'
    mean_length = filtered_df["Exposé des motifs"].str.len().mean()
    length_interval_around_mean_to_keep_ = 500

    # Filter rows where the length of 'Exposé des motifs' is within one standard deviation of the mean
    result_df = filtered_df[
        (
            filtered_df["Exposé des motifs"].str.len()
            >= mean_length - length_interval_around_mean_to_keep_
        )
        & (
            filtered_df["Exposé des motifs"].str.len()
            <= mean_length + length_interval_around_mean_to_keep_
        )
    ]
```

```python
# Shuffle the result_df
shuffled_df = result_df.sample(frac=1).reset_index(drop=True)

for i in range(3):
    line = shuffled_df.iloc[i]
    print(line["Exposé des motifs"])
    print("\n\n")
    ollama_client = SummaryGeneratorOllamaClient()
    prompt = ollama_client.build_prompt(line["Exposé des motifs"])
    summary = ollama_client.generate_response(prompt)
    print("\n\n======= SUMMARY ========\n\n")
    print(f'Expert summary: {line["Objet"]}')
    print(f"LLM summary: {summary}")
    print("\n\n=============================\n\n")
```

```python
# Shuffle the result_df
shuffled_df = result_df.sample(frac=1).reset_index(drop=True)
NB_RESULTS = 50

# Create a list to store the summaries and "Exposé des motifs"
data = []
ollama_client = SummaryGeneratorOllamaClient()

for i in range(NB_RESULTS):
    print(f"i : {i}")
    line = shuffled_df.iloc[i]
    exposé_des_motifs = line["Exposé des motifs"]

    prompt = ollama_client.build_prompt(exposé_des_motifs)
    summary = ollama_client.generate_response(prompt)

    experimental_prompt = ollama_client.build_prompt_experimental(exposé_des_motifs)
    experimental_summary = ollama_client.generate_response(experimental_prompt)

    # Append the data to the list
    data.append(
        {
            "Exposé des motifs": exposé_des_motifs.replace('"', '""').replace('\n', ' '),
            "Objet (Expert)": line["Objet"].replace('"', '""').replace('\n', ' '),
            "Objet (LLaMa 3)": summary.replace('"', '""').replace('\n', ' '),
            "Objet (Exp LLaMa 3)": experimental_summary.replace('"', '""').replace('\n', ' '),
        }
    )

# Create a dataframe from the data
df = pd.DataFrame(data)

# Save the dataframe to a CSV file
df.to_csv(f"{DATA_FOLDER}/échantillons_objets_amendements.csv", index=False)
```
