```python
import logging
import logging.config
import os

import pandas as pd

from graal.summary.llm_clients import FakeLLMAPIClient, LLMAPIClient
from graal.summary.llm_clients import EtalabAPIClient
from graal.summary.summary_prompt_builder import SummaryPromptBuilder
import logging

logging.config.fileConfig("logging.conf")
```

```python
DATA_FOLDER = os.getenv("DATA_FOLDER")
llm_api_client: LLMAPIClient = EtalabAPIClient(
    base_url="https://albert.api.etalab.gouv.fr/v1",
    api_key=os.getenv("ETALAB_API_KEY"),
    model_name="meta-llama/Meta-Llama-3.1-70B-Instruct",
)
# llm_api_client: LLMAPIClient = FakeLLMAPIClient()

df_test = pd.read_csv(os.path.join(DATA_FOLDER, "Fichier de test DFAS.csv"))

df_test["Objet GRAAL"] = df_test["Exposé amdt"].apply(
    lambda value: (
        logging.info(
            f"Generating summary for: {value[:30]}..."
        ),  # Log the first 30 characters of the value
        llm_api_client.generate_summary(SummaryPromptBuilder.build_prompt_dfas(value)),
    )[1]  # Return the result of the summary generation
)
```

```python
import pandas as pd
import re

# Define a function to clean up illegal characters
def remove_illegal_characters(value):
    if isinstance(value, str):
        # Replace all illegal characters with a space or remove them
        return re.sub(r'[\x00-\x1F]+', '', value)
    return value

# Apply the function to all elements in the DataFrame
df_clean = df_test.applymap(remove_illegal_characters)

# Now try saving to Excel
df_clean.to_excel(os.path.join(DATA_FOLDER, "resultat_test_DFAS.xlsx"), index=False)
```
